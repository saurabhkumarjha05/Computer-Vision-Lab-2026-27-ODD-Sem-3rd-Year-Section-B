```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image
image = cv2.imread("images/input.jpg")

if image is None:
    raise FileNotFoundError("Could not find images/input.jpg")

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(8, 5))
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(gray.ravel(), bins=256, range=[0, 256])
plt.title("Original Image Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.show()
```

```python
min_val = np.min(gray)
max_val = np.max(gray)

stretched = ((gray - min_val) / (max_val - min_val) * 255).astype(np.uint8)

plt.figure(figsize=(8, 5))
plt.imshow(stretched, cmap="gray")
plt.title("Contrast Stretched Image")
plt.axis("off")
plt.show()
```

```python
equalized = cv2.equalizeHist(gray)

plt.figure(figsize=(8, 5))
plt.imshow(equalized, cmap="gray")
plt.title("Histogram Equalized Image")
plt.axis("off")
plt.show()
```

```python
clahe = cv2.createCLAHE(
    clipLimit=2.0,
    tileGridSize=(8, 8)
)

clahe_image = clahe.apply(gray)

plt.figure(figsize=(8, 5))
plt.imshow(clahe_image, cmap="gray")
plt.title("CLAHE Enhanced Image")
plt.axis("off")
plt.show()
```

```python
fig, axes = plt.subplots(3, 1, figsize=(10, 12))

axes[0].hist(gray.ravel(), bins=256, range=[0, 256])
axes[0].set_title("Original Histogram")

axes[1].hist(equalized.ravel(), bins=256, range=[0, 256])
axes[1].set_title("Histogram Equalization")

axes[2].hist(clahe_image.ravel(), bins=256, range=[0, 256])
axes[2].set_title("CLAHE Histogram")

plt.tight_layout()
plt.show()
```

```python
# Load the second image
image2 = cv2.imread("images/input2.jpg")

if image2 is None:
    raise FileNotFoundError("Could not find images/input2.jpg")

# Convert to grayscale
gray2 = cv2.cvtColor(image2, cv2.COLOR_BGR2GRAY)

# Display original second image
plt.figure(figsize=(8, 5))
plt.imshow(gray2, cmap="gray")
plt.title("Second Image - Grayscale")
plt.axis("off")
plt.show()
```

## 1. What is image contrast, and why is it important in image processing?

**Answer:**
Image contrast is the difference in intensity between the bright and dark regions of an image. A high-contrast image has a wide range of intensity values, while a low-contrast image has intensity values concentrated within a narrow range.

Contrast is important because it determines how clearly objects, edges, and details can be distinguished. Improving contrast makes image features more visible and can help subsequent computer vision tasks such as feature extraction, segmentation, object detection, and recognition.

---

## 2. Explain the concept of an image histogram. What information does it provide?

**Answer:**
An image histogram is a graphical representation of the distribution of pixel intensity values in an image. For a grayscale image, the horizontal axis represents intensity values, usually from **0 to 255**, while the vertical axis represents the number of pixels having each intensity.

A histogram provides information about the **brightness and contrast** of an image. It can show whether an image is predominantly dark, bright, or has a narrow or wide distribution of intensity values. Histogram analysis is useful for determining whether contrast enhancement is required.

## 3. Differentiate between Histogram Stretching and Histogram Equalization.

**Answer:**

| Histogram Stretching                                                             | Histogram Equalization                                                                              |
| -------------------------------------------------------------------------------- | --------------------------------------------------------------------------------------------------- |
| Expands the existing intensity range of an image.                                | Redistributes pixel intensities to produce a more uniform histogram.                                |
| Usually maps the minimum and maximum intensities to the full available range.    | Uses the cumulative distribution of pixel intensities for redistribution.                           |
| Preserves the general distribution of intensities more directly.                 | Can significantly change the intensity distribution.                                                |
| Useful when an image uses only a small portion of the available intensity range. | Useful when pixel intensities are unevenly distributed and stronger contrast enhancement is needed. |

In this experiment, **contrast stretching** was used to improve the dynamic range, while **Histogram Equalization** was used to redistribute intensity values and enhance contrast.

## 4. What is Contrast Limited Adaptive Histogram Equalization (CLAHE)? How does it differ from standard Histogram Equalization?

**Answer:**
CLAHE stands for **Contrast Limited Adaptive Histogram Equalization**. It is a local contrast enhancement technique that divides an image into small regions, called tiles, and performs histogram equalization separately in those regions.

The main difference is that standard Histogram Equalization operates on the **whole image globally**, whereas CLAHE works on **local regions**. CLAHE also limits excessive contrast amplification using a clip limit. As a result, CLAHE can improve local details while reducing the risk of over-enhancing noise.

This makes CLAHE particularly useful for images with **varying illumination**, as required in this experiment.

## 5. Why is histogram equalization commonly applied before feature extraction and image segmentation?

**Answer:**
Histogram equalization is commonly applied before feature extraction and segmentation because it can improve image contrast and make important features more visible.

When the intensity distribution is poor or concentrated in a narrow range, edges and object boundaries may be difficult to distinguish. Histogram equalization can spread the intensity values over a wider range, making these structures easier to detect.

Therefore, improved contrast can provide better input for subsequent computer vision operations such as **feature extraction and image segmentation**.

## 7. What are the limitations of global histogram equalization?

**Answer:**
Global Histogram Equalization has several limitations:

* It treats the **entire image as one region**, so it cannot adapt to local variations in illumination.
* It may produce **excessive contrast** in some areas.
* It can make **noise more visible**, particularly in nearly uniform regions.
* It may produce unnatural-looking results when different parts of an image have very different brightness levels.

Because of these limitations, local methods such as **CLAHE** can be more suitable for images with varying illumination.

## 9. Compare histogram-based enhancement techniques with brightness adjustment methods.

**Answer:**

| Histogram-Based Enhancement                                             | Brightness Adjustment                                                   |
| ----------------------------------------------------------------------- | ----------------------------------------------------------------------- |
| Changes the distribution of pixel intensities to improve contrast.      | Primarily shifts pixel intensities to make an image brighter or darker. |
| Can improve both overall and local contrast depending on the technique. | Does not necessarily improve contrast.                                  |
| Includes techniques such as Histogram Equalization and CLAHE.           | Usually involves adding or subtracting a fixed intensity value.         |
| Can reveal details hidden due to poor contrast.                         | May make an image brighter while leaving low contrast unchanged.        |

Therefore, histogram-based techniques are generally more effective when the main problem is **poor contrast**, whereas brightness adjustment is more appropriate when the main problem is simply that the image is too dark or too bright.

## 10. Why is CLAHE preferred for medical imaging and low-light image enhancement?

**Answer:**
CLAHE is preferred because it performs **local contrast enhancement** instead of applying the same transformation to the entire image. This allows it to improve details in different regions even when illumination varies across the image.

CLAHE also uses a **contrast limit** to reduce excessive contrast amplification, which can help prevent unwanted enhancement of noise. Therefore, it is particularly useful for medical and low-light images where preserving local details and handling uneven illumination are important.
